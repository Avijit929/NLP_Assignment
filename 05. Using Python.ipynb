1. Fixed Stack Dynamic

Allocation: Stack

Size: Fixed and known at compile-time

```python
# A list with a fixed size inside a function, initialized directly

def fixed_stack_dynamic():
    arr = [0] * 5  # Fixed size at definition
    for i in range(len(arr)):
        arr[i] = i * 10
    print("Fixed Stack Dynamic:", arr)

fixed_stack_dynamic()
```

2. Stack Dynamic

Allocation: Stack

Size: Determined at runtime

```python
# A list size based on user input or a calculated value inside a function

def stack_dynamic():
    size = int(input("Enter size of array: "))
    arr = [i for i in range(size)]  # Dynamic size, allocated during execution
    print("Stack Dynamic:", arr)

stack_dynamic()
```

3. Fixed Heap Dynamic

Allocation: Heap

Size: Fixed during object creation

```python
# Class with a list created in the constructor of fixed size

class FixedHeapDynamic:
    def __init__(self):
        self.arr = [i * 2 for i in range(5)]  # Fixed size once object is created

    def display(self):
        print("Fixed Heap Dynamic:", self.arr)

obj = FixedHeapDynamic()
obj.display()
```

4. Heap Dynamic

Allocation: Heap

Size: Can change at runtime

```python
# Dynamic list using append, pop

def heap_dynamic():
    arr = []  # Starts empty, heap allocated
    for i in range(5):
        arr.append(i * 3)
    arr.append(100)    # Grow
    arr.pop(2)         # Shrink
    print("Heap Dynamic:", arr)

heap_dynamic()
```
